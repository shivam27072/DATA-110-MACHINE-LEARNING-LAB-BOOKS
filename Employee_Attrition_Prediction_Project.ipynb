# Data Preprocessing

B**efore training machine learning models, the dataset is preprocessed by removing irrelevant features, encoding categorical variables, splitting the dataset into training and testing sets, and scaling numerical features where required.**

```python
# Remove unnecessary columns

df.drop(
    columns=[
        "EmployeeCount",
        "EmployeeNumber",
        "Over18",
        "StandardHours"
    ],
    inplace=True
)
```

The columns EmployeeCount, EmployeeNumber, Over18, and StandardHours were removed because they either contain constant values or unique identifiers that do not contribute to predicting employee attrition.

```python
# Features and Target

X = df.drop("Attrition", axis=1)
y = df["Attrition"]
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y)
```

```python
X.select_dtypes(include="object").columns
```

```python
X = pd.get_dummies(X, drop_first=True)
```

**Categorical variables were converted into numerical format using one-hot encoding so that machine learning algorithms could process them effectively.**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train.shape)
print(X_test.shape)
```

# Model Building

**In this section, three supervised machine learning classification algorithms are trained and evaluated to predict employee attrition. Their performance is compared using multiple evaluation metrics such as Accuracy, Precision, Recall, and F1-Score.**

```python
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
log_model = LogisticRegression(random_state=42)
```

```python
log_model.fit(X_train_scaled, y_train)
```

```python
y_pred_log = log_model.predict(X_test_scaled)
```

```python
log_accuracy = accuracy_score(y_test, y_pred_log)
log_precision = precision_score(y_test, y_pred_log)
log_recall = recall_score(y_test, y_pred_log)
log_f1 = f1_score(y_test, y_pred_log)
```

```python
print("========== Logistic Regression ==========\n")

print(f"Accuracy : {log_accuracy:.4f}")
print(f"Precision: {log_precision:.4f}")
print(f"Recall   : {log_recall:.4f}")
print(f"F1 Score : {log_f1:.4f}")
```

```python
print("\nClassification Report\n")

print(classification_report(y_test, y_pred_log))
```

```python
plt.figure(figsize=(5,4))

cm = confusion_matrix(y_test, y_pred_log)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Logistic Regression Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")

plt.show()
```

### Logistic Regression Observation

**Logistic Regression was used as the baseline classification model for predicting employee attrition. The model demonstrated good predictive performance and provides a benchmark for comparison with Decision Tree and Random Forest models.**
