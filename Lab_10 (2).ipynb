```python
# Sample Dataset
data = {
    "Age": [20, 21, 22, 23, 24, 25, 26],
    "Marks": [85, np.nan, 78, 90, np.nan, 88, 95],
    "City": ["Delhi", "Mumbai", np.nan, "Delhi", "Chennai", "Mumbai", np.nan]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Original Dataset\n")
print(df)
```

```python
# Drop rows with missing values
df_drop_rows = df.dropna()

print("\nRows after dropping missing values")
print(df_drop_rows)
```

```python
# Drop columns with missing values
df_drop_col = df.dropna(axis=1)

print("\nColumns after dropping missing values")
print(df_drop_col)
```

```python
# Fill missing values with a constant
df_constant = df.fillna(99999)

print("\nFill with Constant (99999)")
print(df_constant)
```

```python
# Fill missing values with NA
df_constant = df.fillna("NA")

print(df_constant)
```

```python
# Mean Imputation
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())
df_mean["Marks"] = df_mean["Marks"].fillna(df_mean["Marks"].mean())

print("\nMean Imputation")
print(df_mean)
```

```python
# Median Imputation
df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())
df_median["Marks"] = df_median["Marks"].fillna(df_median["Marks"].median())

print("\nMedian Imputation")
print(df_median)
```

```python
# Mode Imputation
df_mode = df.copy()
df_mode["City"] = df_mode["City"].fillna(df_mode["City"].mode()[0])

print("\nMode Imputation")
print(df_mode)
```

```python
# Forward Fill
df_ffill = df.ffill()
print("\nForward Fill")
print(df_ffill)
```

```python
# Backward Fill
df_bfill = df.bfill()
print("\nBackward Fill")
print(df_bfill)
```

```python
# Linear Intercolation
df_interpolate = df.interpolate()
df_interpolate["Age"] = df_interpolate["Age"].interpolate()
df_interpolate["Marks"] = df_interpolate["Marks"].interpolate()
print("\nLinear Intercolation")
print(df_interpolate)
```

```python
# Specific Value Imputation
```

```python
# Random Imputation
df_random = df.copy()

for col in ["Age", "Marks"]:
    missing = df_random[col].isnull()
    random_values = df_random[col].dropna().sample(missing.sum(), replace=True).values
    df_random.loc[missing, col] = random_values

print("\nRandom Imputation")
print(df_random)
```
